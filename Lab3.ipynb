```python
def minimax(node, is_max):
    """
    Recursive function to compute the minimax value of a node.

    Parameters:
        node (dict or int): Current node in the game tree.
        is_max (bool): True if the current player is MAX, False if MIN.

    Returns:
        int: Optimal value for the current player.
    """
    # Caso base: si el nodo es un valor entero (hoja)
    if isinstance(node, int):
        return node

    # Si es el turno de MAX, maximiza el valor
    if is_max:
        best_value = float('-inf')
        for child in node.values():
            best_value = max(best_value, minimax(child, False))
        return best_value
    # Si es el turno de MIN, minimiza el valor
    else:
        best_value = float('inf')
        for child in node.values():
            best_value = min(best_value, minimax(child, True))
        return best_value

# Escenario 1: Árbol de juego básico
game_tree = {
    "root": {"A": {"C": 3, "D": 5}, "B": {"E": 6, "F": 9}}
}
optimal_value = minimax(game_tree["root"], is_max=True)
print(f"Optimal value for MAX (Basic Tree): {optimal_value}")
# Resultado esperado: 5

# Escenario 2: Árbol con valores iguales
game_tree = {
    "root": {"A": {"C": 4, "D": 4}, "B": {"E": 4, "F": 4}}
}
optimal_value = minimax(game_tree["root"], is_max=True)
print(f"Optimal value for MAX (Equal Values Tree): {optimal_value}")
# Resultado esperado: 4

# Escenario 3: MAX con movimiento ganador inmediato
game_tree = {
    "root": {"A": 10, "B": {"C": 3, "D": 5}}
}
optimal_value = minimax(game_tree["root"], is_max=True)
print(f"Optimal value for MAX (Immediate Win Tree): {optimal_value}")
# Resultado esperado: 10

# Escenario 4: Árbol asimétrico simple
game_tree = {
    "root": {"A": {"C": 3}, "B": {"D": 5, "E": 2}}
}
optimal_value = minimax(game_tree["root"], is_max=True)
print(f"Optimal value for MAX (Asymmetric Tree): {optimal_value}")
# Resultado esperado: 5
```

1. What happens when all terminal nodes have the same utility value?

When all terminal nodes have the same value, the Minimax algorithm cannot identify any advantage for either player. Both the MAX and MIN players will select any valid branch, as they all yield the same result.
In this case, the final outcome will always be the common value of the terminal nodes.


2. How does the algorithm behave if MAX always has an immediate winning move?

If the MAX player has an immediate winning move, the Minimax algorithm will directly select that move. This happens because the algorithm always prioritizes maximizing the value at each decision level for MAX.
MAX will choose the branch with the highest available value without needing to explore other paths.

3. Explain the role of recursion in the Minimax algorithm.

Recursion is fundamental in the Minimax algorithm because it systematically explores all possible future decisions.
At each level of the tree, the algorithm alternates between MAX and MIN players.
The algorithm dives down to terminal nodes (base case).
Then, it returns the evaluated values upwards, allowing each player to make the best possible decision on their turn.

Analogy: It’s like a family tree where every possible descendant is evaluated to determine the optimal outcome.
Benefit: Recursion ensures that no branch of the tree is omitted and guarantees an optimal result, assuming perfect play from both players.

```python
# Escenario 3: Implementación de Alpha-Beta Pruning (usando el árbol de ejemplo proporcionado)
game_tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [], 'E': [], 'F': [], 'G': []
}

# Valores terminales para las hojas
terminal_values = {
    'D': 3, 'E': 5,
    'F': 6, 'G': 9
}

def alpha_beta_pruning(node, depth, alpha, beta, maximizing_player):
    """
    Implementa Alpha-Beta Pruning para calcular el valor óptimo.

    Parámetros:
        node (str): Nodo actual en el árbol de juego.
        depth (int): Profundidad actual en el árbol.
        alpha (float): Mejor opción explorada para el jugador MAX.
        beta (float): Mejor opción explorada para el jugador MIN.
        maximizing_player (bool): True si es el turno del jugador MAX.

    Retorna:
        int: Valor óptimo para el jugador actual.
    """
    if node in terminal_values:
        return terminal_values[node]

    if maximizing_player:
        max_eval = float('-inf')
        for child in game_tree[node]:
            eval = alpha_beta_pruning(child, depth + 1, alpha, beta, False)
            max_eval = max(max_eval, eval)
            alpha = max(alpha, eval)
            if beta <= alpha:
                print(f"Poda Beta en nodo {child} con alpha={alpha} y beta={beta}")
                break  # Beta cut-off
        return max_eval
    else:
        min_eval = float('inf')
        for child in game_tree[node]:
            eval = alpha_beta_pruning(child, depth + 1, alpha, beta, True)
            min_eval = min(min_eval, eval)
            beta = min(beta, eval)
            if beta <= alpha:
                print(f"Poda Alpha en nodo {child} con alpha={alpha} y beta={beta}")
                break  # Alpha cut-off
        return min_eval

optimal_value = alpha_beta_pruning('A', 0, float('-inf'), float('inf'), True)
print(f"El valor óptimo usando Alpha-Beta Pruning (Árbol de ejemplo): {optimal_value}")
```
